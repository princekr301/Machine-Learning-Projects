```python
s = {'setosa':0, 'versicolor':1, 'virginica':2}
```

```python
df["species"] = df["species"].map(s)
```

```python
df.sample(5)
```

```python
x = df.drop("species",axis=1)
y = df[["species"]]
```

```python
x.head()
```

```python
y.head()
```

```python
df.shape
```

```python
# How to Divide the data for training and testing
from sklearn.model_selection import train_test_split
```

```python
# 100% =   Train  80%      Test  20%
#             x,y           x,y

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
# Build a Machine learning Model Which is used for classification
# Logistic Regression :  
```

```python
log = LogisticRegression()
log.fit(x_train,y_train)
```

```python
pred = log.predict(x_test)
print(pred)
```

```python
s
```

```python
ac = log.score(x_train,y_train) # Show the Accuracy of training
print(ac)
```

```python
ac2 = log.score(x_test,y_test) # Show the accuracy of testing data
print(ac2)
```
